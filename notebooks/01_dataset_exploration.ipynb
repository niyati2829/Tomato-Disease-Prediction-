### Step 1: Environment & Path Setup
We import our modular functions from `src.data.explore_dataset` to keep our code clean, reusable, and maintainable.

```python
from pathlib import Path
import sys

# Ensure project root is in sys.path
project_root = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(project_root) not in sys.path:
    sys.path.append(str(project_root))

from src.data.explore_dataset import (
    find_tomato_classes,
    count_and_verify_images,
    inspect_image_properties,
    analyze_class_imbalance,
    plot_class_distribution,
    plot_sample_grid,
    run_exploration,
)

color_dir = project_root / "data" / "raw" / "PlantVillage-Dataset" / "raw" / "color"
figures_dir = project_root / "artifacts" / "figures"
print("Project root:", project_root)
print("Dataset path:", color_dir)
```

### Step 2: Locate Classes & Verify Image Integrity
We scan the directories to confirm all 10 tomato categories and verify that every image can be decoded without corruption.

```python
tomato_dirs = find_tomato_classes(color_dir)
print(f"Found {len(tomato_dirs)} Tomato class folders:")
for name in tomato_dirs.keys():
    print(f"  - {name}")

df, corrupt_files = count_and_verify_images(tomato_dirs)
print(f"\nTotal Verified Images: {df['count'].sum():,}")
print(f"Corrupt Images Found: {len(corrupt_files)}")
df
```

### Step 3: Inspect Image Properties
Here we verify that all images share standard dimensions (256x256), RGB color channels, and JPEG encoding.

```python
props = inspect_image_properties(tomato_dirs, sample_per_class=15)
print("Sampled Image Properties:")
print(" - Unique Dimensions (W, H):", props["unique_dimensions_wh"])
print(" - Unique Color Modes:", props["unique_color_modes"])
print(" - Unique File Formats:", props["unique_file_formats"])
```

### Step 4: Class Imbalance Analysis
We measure the ratio between the majority class and minority class to inform our splitting and loss function strategies.

```python
imbalance = analyze_class_imbalance(df)
print(f"Majority Class: {imbalance['majority_class']} ({imbalance['majority_count']:,} images)")
print(f"Minority Class: {imbalance['minority_class']} ({imbalance['minority_count']:,} images)")
print(f"Imbalance Ratio: {imbalance['imbalance_ratio']:.2f}x")
```

### Step 5: Visualize Class Distribution & Representative Leaves
Save the generated plots to `artifacts/figures/` and display them inline.

```python
dist_fig = figures_dir / "tomato_class_distribution.png"
sample_fig = figures_dir / "tomato_sample_grid.png"

plot_class_distribution(df, dist_fig)
plot_sample_grid(tomato_dirs, sample_fig, seed=42)

from IPython.display import Image as IPImage, display
display(IPImage(filename=str(dist_fig)))
display(IPImage(filename=str(sample_fig)))
```
